## 2.2 Strings

A **string** is a sequence of characters (letters, digits, spaces, symbols) enclosed in quotes. In Python you can use single quotes `'...'` or double quotes `"..."`.

Strings are used everywhere in music analysis: pitch names, part names, file paths, tempo markings, key names.

````python
pitch_name = "C#"
key_signature = 'D minor'
file_path = "bach/bwv269"

# Strings can be combined with + (called concatenation)
description = "The piece is in " + key_signature
print(description)

# Strings have a length
print("Number of characters in the key name:", len(key_signature))

# You can check if something is inside a string
print("minor" in key_signature)   # prints True or False
````

## 2.3 Lists

A **list** is an ordered collection of items, written with square brackets `[item1, item2, item3]`. Lists are very common in music analysis — a sequence of pitches, a list of note durations, the parts of a score.

````python
# A list of pitch names
c_major_scale = ["C", "D", "E", "F", "G", "A", "B"]

# Access individual items by their position (counting starts at 0!)
print("First note:", c_major_scale[0])
print("Third note:", c_major_scale[2])
print("Last note:", c_major_scale[-1])   # -1 means the last item

# How many items are in the list?
print("Number of notes:", len(c_major_scale))
````

## 2.4 Functions

A **function** is a reusable block of code that performs a specific task. You *call* (use) a function by writing its name followed by parentheses, with any inputs (called **arguments**) inside:

```python
result = function_name(argument1, argument2)
```

You have already used functions: `print()` and `len()` are both built-in Python functions.

Functions can also belong to objects — you access them with a dot:

```python
result = some_object.method_name(argument)
```

This is called a **method**. In music21 you will see this pattern constantly: `score.show()`, `note.pitch.name`, `stream.getElementsByClass()`.

````python
# Built-in functions
pitches = ["G", "A", "B", "C", "D"]
print(len(pitches))         # len() counts items
print(sorted(pitches))      # sorted() returns an alphabetically sorted list

# A method called on a string object
key_name = "d minor"
print(key_name.upper())     # .upper() converts to uppercase
print(key_name.replace("d", "G"))  # .replace() substitutes text
````

## 2.6 Loops

A **loop** repeats a block of code multiple times. The most common type in Python is the `for` loop, which goes through each item in a collection:

```python
for item in collection:
    do_something_with(item)
```

**The indentation matters!** The code inside the loop must be indented (shifted right, usually by 4 spaces). Python uses indentation to know where the loop body begins and ends.

````python
# Loop through a scale and print each pitch
c_major = ["C", "D", "E", "F", "G", "A", "B"]

for pitch in c_major:
    print("Pitch:", pitch)

print("--- done ---")
````

````python
# Loops can also build up a result
durations = [1.0, 0.5, 0.5, 1.0, 2.0]   # in quarter note units
total_duration = 0

for d in durations:
    total_duration = total_duration + d   # add each duration to the running total

print("Total duration in quarter notes:", total_duration)
````

## 3.2 The Built-in Corpus

Music21 comes with a large collection of scores already included — this is called the **corpus**. It contains hundreds of works: Bach chorales, Mozart piano sonatas, Beethoven string quartets, Josquin masses, and folk songs from many traditions.

A full list is available at: https://web.mit.edu/music21/doc/about/referenceCorpus.html

To load a score from the corpus we use `corpus.parse()` and give it a path or identifier.

````python
import music21
print("Music21 version:", music21.__version__)
print("Music21 loaded successfully!")
````

````python
# Load a Bach chorale from the built-in corpus
# BWV 269 is 'Ach wie nichtig, ach wie flüchtig'
chorale = music21.corpus.parse('bach/bwv269')

# What type of object did we get?
print(type(chorale))
print("Score loaded!")
````

## 3.3 Exploring the Score Structure

A score in music21 is a hierarchical object. Think of it like a set of nested containers:

```
Score
  └── Part (e.g. Soprano, Alto, Tenor, Bass)
        └── Measure (bar)
              └── Note / Rest / Chord
```

We can inspect the top level of the score to see its parts.

````python
# Show the title and parts of the score
print("Title:", chorale.metadata.title)
print("Composer:", chorale.metadata.composer)
print()

# List all the parts (voices)
print("Parts in the score:")
for part in chorale.parts:
    print(" -", part.id)
````

````python
# How many measures does the chorale have?
soprano_part = chorale.parts['Soprano']
measures = soprano_part.getElementsByClass('Measure')
print("Number of measures:", len(list(measures)))
````

## 3.4 Displaying Sheet Music


One of the most satisfying features of music21 is being able to display actual sheet music inside the notebook. We use the `.show()` method.



````python
# Display the first 4 measures of the soprano part
# .measures(start, end) extracts a range of bars
excerpt = soprano_part.measures(1, 4)
excerpt.show('text')   # 'text' mode shows a text representation — useful for quick inspection
````

````python
# Display as graphical notation (this opens a score image in the output)
# If this doesn't work in your Colab session, use show('text') as above
excerpt.show()
````

## 3.5 Extracting Notes

We can pull individual notes from the score and inspect their properties. Each note object in music21 has many attributes: `pitch.name`, `pitch.octave`, `duration.quarterLength`, `offset` (position in time), and more.

````python
# Get the soprano part
soprano = chorale.parts['Soprano']

# Extract all notes from the soprano (not rests)
soprano_notes = soprano.flatten().notes   # .flatten "flattens" the hierarchy so we can loop through all notes

# Print the first 10 notes
print("First 10 soprano notes:")
count = 0
for note in soprano_notes:
    if count < 10:
        print(f"  Pitch: {note.pitch.name}, Octave: {note.pitch.octave}, Duration: {note.duration.quarterLength}")
        count = count + 1
````

## 3.6 Calculating Something: Pitch Counts

Now let's ask a real analytical question: **Which pitch classes appear most frequently in the soprano part?**

A *pitch class* is the name of a pitch without the octave information (so C4 and C5 are both just "C").

````python
# Count how many times each pitch class appears in the soprano
from collections import Counter   # Counter is a helpful tool for counting

soprano = chorale.parts['Soprano']
soprano_notes = soprano.flat.notes

# Build a list of all pitch names
pitch_names = []
for note in soprano_notes:
    pitch_names.append(note.pitch.name)   # .append() adds an item to a list

# Count how often each pitch name occurs
pitch_counts = Counter(pitch_names)

# Print the counts sorted from most to least frequent
print("Pitch frequencies in the Soprano:")
for pitch, count in pitch_counts.most_common():
    print(f"  {pitch}: {count} times")
````

````python
# Music21 can also estimate the key automatically
estimated_key = chorale.analyze('key')
print("Estimated key:", estimated_key)
print("Tonic:", estimated_key.tonic.name)
print("Mode:", estimated_key.mode)
print("Confidence:", round(estimated_key.correlationCoefficient, 3))
````
